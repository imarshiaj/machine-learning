```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

digits = load_digits()
fig, axes = plt.subplots(1, 4, figsize=(8,2))
for ax, image, label in zip(axes, digits.images[:4], digits.target[:4]):
    ax.imshow(image, cmap="gray_r")
    ax.set_title(str(label)); ax.axis("off")
plt.suptitle("Computer vision: image pixels and labels")
plt.tight_layout()
```

## Modality-specific considerations
- **Computer vision:** image resize/color conventions, augmentation, duplicate/near-duplicate leakage, patient/person-level splits, class imbalance, and localization metrics for detection.
- **NLP:** tokenization, language coverage, prompt/context length, class imbalance, annotation ambiguity, and privacy in text.
- **Audio/speech:** sample rate, segmentation, silence/noise, speaker overlap, word error rate for transcription, and speaker-independent splits.

Transfer learning is common because a pretrained representation can reduce the amount of labeled data required, but the training data and license of the pretrained model still matter.
