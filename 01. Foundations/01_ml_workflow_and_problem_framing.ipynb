## Learning goals

Machine learning is not simply fitting an estimator. A useful workflow begins by asking what decision the prediction will support, what information is available at prediction time, and how success will be measured. This notebook uses small examples and avoids assuming that a model is automatically useful.

## 1. Match the question to a task

| Question | Typical task | Target / output | Example metric |
|---|---|---|---|
| How much will a home cost? | Regression | Continuous number | MAE / RMSE |
| Will a customer leave? | Binary classification | Class or probability | Recall, PR-AUC, calibration |
| Which customers behave similarly? | Clustering | Group assignment | Stability + domain review |
| Which items should appear first? | Ranking | Ordered list | NDCG / business outcome |

The metric must reflect the cost of errors. Accuracy alone is often misleading when classes are imbalanced.

```python
# A simple problem-framing template
problem = {
    "decision": "Which customers should receive a retention offer?",
    "unit_of_analysis": "one customer per prediction date",
    "target": "churn within the next 30 days",
    "prediction_time": "the end of the current month",
    "available_features": ["tenure", "contract type", "recent support contacts"],
    "primary_metric": "recall at a fixed outreach budget",
    "baseline": "predict no churn, then compare with a simple rule",
    "constraints": ["avoid future information", "respect customer privacy"],
}
for key, value in problem.items():
    print(f"{key:20}: {value}")
```

## 2. Separate the lifecycle into explicit stages

1. Clarify the problem and prediction unit.
2. Collect and audit data.
3. Split data before learning preprocessing parameters.
4. Build a baseline and a reproducible pipeline.
5. Tune using validation / cross-validation only.
6. Evaluate once on a held-out test set.
7. Deploy, log predictions, monitor changes, and decide when to retrain.

**Important:** a good offline score does not guarantee business value. Check data availability, latency, interpretability needs, failure costs, and how a human will use the output.

```python
# A baseline matters because it answers: "Better than what?"
import numpy as np
from sklearn.metrics import accuracy_score

# Toy binary outcome: only 8% of cases are positive.
y_true = np.array([0] * 92 + [1] * 8)
y_naive = np.zeros_like(y_true)
print(f"Always-negative baseline accuracy: {accuracy_score(y_true, y_naive):.1%}")
print("Yet the baseline detects none of the positive cases.")
```

## Common pitfalls
- Choosing the metric after seeing test results.
- Defining the target in a way that is not actionable.
- Using variables that would not be known when the prediction is made.
- Ignoring a simple baseline.
- Treating correlation as proof that an intervention will work.
